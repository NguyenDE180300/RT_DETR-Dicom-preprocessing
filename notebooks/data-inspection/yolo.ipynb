```python
import json

path = "/home/Ubuntu/Lung_2026/vinbigdata-CLAHE-png/annotations/instances_train_mmdet.json"

with open(path, "r") as f:
    coco = json.load(f)

print("images:", len(coco["images"]))
print("annotations:", len(coco["annotations"]))
print("categories:", len(coco["categories"]))

print("\nCATEGORY NAMES:")
for c in coco["categories"]:
    print(c["id"], c["name"])

# check id continuity
ids = [c["id"] for c in coco["categories"]]
print("\ncategory_id sorted:", sorted(ids))
```

```python
bad = 0

for ann in coco["annotations"]:
    x, y, w, h = ann["bbox"]
    if w <= 0 or h <= 0:
        bad += 1

print("Invalid bbox:", bad)
```

```python
image_ids = set(img["id"] for img in coco["images"])
ann_ids = set(ann["image_id"] for ann in coco["annotations"])

print("Missing images:", len(ann_ids - image_ids))
print("Orphan images:", len(image_ids - ann_ids))
```

```python
import json

path = "/home/Ubuntu/Lung_2026/vinbigdata-CLAHE-png/annotations/instances_test.json"

with open(path, "r") as f:
    coco = json.load(f)

# fix category ids
id_map = {}

for i, cat in enumerate(coco["categories"]):
    old_id = cat["id"]
    new_id = i
    id_map[old_id] = new_id
    cat["id"] = new_id

# fix annotations
for ann in coco["annotations"]:
    ann["category_id"] = id_map[ann["category_id"]]

# save new file
out_path = path.replace(".json", "_mmdet.json")

with open(out_path, "w") as f:
    json.dump(coco, f)

print("DONE:", out_path)
```
